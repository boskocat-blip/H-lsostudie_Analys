Syfte: Jag vill se om det finns skillnad i systoliskt blodtryck mellan rökare och icke-rökare.  
Jag använder ett t-test för att jämföra grupperna.


```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

# läser in filen (ska ligga i samma mapp)
data = pd.read_csv("health_study_dataset.csv")

# kollar lite på datan
print("Första rader:")
print(data.head())

print("\nKolumner:")
print(data.columns.tolist())
```

```python
# bara för att se lite statistik
print("Statistik för blodtryck, vikt och ålder:")
print(data[["age", "weight", "systolic_bp"]].describe())

# enkel graf för blodtryck
plt.hist(data["systolic_bp"], bins=15, color="skyblue", edgecolor="black")
plt.title("Histogram – Systoliskt blodtryck")
plt.xlabel("mmHg")
plt.ylabel("Antal personer")
plt.show()
```

```python
# delar upp i rökare och icke-rökare
rok = data[data["smoker"] == "Yes"]["systolic_bp"].dropna()
inte_rok = data[data["smoker"] == "No"]["systolic_bp"].dropna()

# gör t-test (Welch)
t, p = stats.ttest_ind(rok, inte_rok, equal_var=False)

print("t-värde:", round(t, 3))
print("p-värde:", round(p, 4))
print("Medel rökare:", round(rok.mean(), 2))
print("Medel icke-rökare:", round(inte_rok.mean(), 2))

if p < 0.05:
    print("\n--> Det verkar finnas en skillnad mellan grupperna.")
else:
    print("\n--> Ingen signifikant skillnad enligt testet.")
```
