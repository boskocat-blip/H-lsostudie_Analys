Simulering + Histogram

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# läsa in datan.
df = pd.read_csv("health_study_dataset.csv")

disease = df["disease"]
verklig_andel = disease.mean()
print("Verklig andel sjukdom i datasetet:", verklig_andel)

#simulera 1000 personer med samma sannolikhet
np.random.seed(7) 

# skapa 1000 slumpvärden mellan 0 och 1
slump = np.random.rand(1000)

# om slumpvärde verklig_andel sjuk
sim = (slump < verklig_andel)

sim_andel = sim.mean()
print("Simulerad andel (1000 personer):", sim_andel)

skillnad = sim_andel - verklig_andel
print("Skillnad mellan simulerad och verklig:", skillnad)

# skapa histogram över simuleringen
# 1 = sjuk, 0 = frisk
plt.hist(sim.astype(int), bins=[-0.5, 0.5, 1.5])
plt.title("Histogram över simulerade sjukdomsfall")
plt.xlabel("Sjukdom (0 = nej, 1 = ja)")
plt.ylabel("Antal personer")
plt.grid(True)
plt.show()
```

Har finns en variabel disease som visar om en person har en sjukdom (1) eller inte (0). För att förstå hur stor andelen sjuka är i populationen gjorde jag först en enkel beräkning av den verkliga andelen i datan:

Verklig andel sjukdom = antal med disease = 1 / totala antal personer

Denna andel använde jag sedan som sannolikhet när jag simulerade 1000 slumpade personer med numpy.random. Tanken är att varje person har samma chans att få sjukdom som genomsnittet i datasetet. På så sätt kan man jämföra om simuleringen ger ett rimligt värde som liknar den verkliga observationen.

Stegen jag gjorde:

-Beräknade verklig andel sjuka i datasetet

-Satte ett seed (så jag får samma resultat varje gång)

-Genererade 1000 slumpvärden mellan 0 och 1

-Om slumpvärdet < sannolikheten = personen klassas som sjuk

-Räknade den simulerade andelen

-Jämförde simulerad andel med verklig

Simuleringen gav ett resultat som ligger ganska nära den verkliga andelen, vilket är vad man förväntar sig när sannolikheten är korrekt angiven.

Jag ritade även ett histogram som visar hur många av de simulerade personerna som blev sjuka (1) respektive friska (0). Histogrammet hjälper att visuellt bekräfta att majoriteten är friska, vilket stämmer med datan.
